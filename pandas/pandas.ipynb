Pandas Series 

A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collevtively called the index.

Creating Series 

There are multiple ways to create a pandas Series.

```python
import numpy as np
import pandas as pd
```

```python
labels = ["a","b","c"]
my_list = [10,20,30]
arr = np.array([10,20,30])
d =  {1:10,2:20,3:30}
```

```python
pd.Series(my_list,index = labels)
```

```python
pd.Series(arr)
```

```python
pd.Series(d)
```

Data Frames

Creating a Dataframes

```python
data = {
    'Name': ['John', 'Anna', 'Peter', 'Linda'],
    'Age': [28, 34, 29, 42],
    'City': ['New York', 'Paris', 'Berlin', 'London'],
    'Salary': [65000, 70000, 62000, 85000]
}
df = pd.DataFrame(data)
df
```

```python

data_list = [
    ['John', 28, 'New York', 65000],
    ['Anna', 34, 'Paris', 70000],
    ['Peter', 29, 'Berlin', 62000],
    ['Linda', 42, 'London', 85000]
]
df2 = pd.DataFrame(data_list)
columns = ["Name","Age","City","Salary"]
df2 = pd.DataFrame(data_list,columns = columns)
df2
```

Selection and indexing of Columns

```python
df2[["Name","City"]]
```

Creating a new columns

```python
df2["Designation"] = ["Doctor","Eng.","Doctor","Eng."]
```

```python
df2
```

Removing Columns

```python
df2 = df2.drop("Designation",axis =1)
df2
```

```python
df2.drop(["City","Salary"],axis =1)
```

```python
df2.drop(0,axis = 0)
```

Conditional Selection

```python
df2
```

```python
# I only want to seem those people whose afe is above 30
df2[df2["Age"] > 30]
```

```python
#I only want poeple whose age is above 30 and their city must be paris
df2[(df2["Age"] > 30) & (df2["City"] == "Paris")]
```
